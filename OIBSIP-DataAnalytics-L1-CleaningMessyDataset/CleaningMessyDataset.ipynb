### **Step 25 — Handle invalid ages**

## Invalid Age Handling

Suspect and victim ages outside the valid human age range of 0–100 were treated as invalid data-entry values.

Rather than deleting the complete incident record, invalid ages were converted to missing values because the remaining information in those incident records may still be useful.

The resulting missing ages are handled using median imputation because age is numeric and the median is less affected by extreme values.

```python
# Convert logically invalid ages to NaN
df.loc[
    (df["suspect_age"] < 0) | (df["suspect_age"] > 100),
    "suspect_age"
] = np.nan

df.loc[
    (df["victim_age"] < 0) | (df["victim_age"] > 100),
    "victim_age"
] = np.nan
```

```python
print("Invalid suspect ages remaining:",
      ((df["suspect_age"] < 0) | (df["suspect_age"] > 100)).sum())

print("Invalid victim ages remaining:",
      ((df["victim_age"] < 0) | (df["victim_age"] > 100)).sum())
```

### **Step 26 — Handle invalid coordinates**

## Geographic Coordinate Cleaning

Latitude values outside -90 to 90 and longitude values outside -180 to 180 were considered invalid.

Invalid coordinate values were converted to missing values rather than deleting the complete incident record because other incident information may still be valid.

Missing coordinates are subsequently imputed using the median coordinate for the corresponding city, with the overall median used as a fallback.

```python
# Replace invalid latitude values with NaN
df.loc[
    (df["latitude"] < -90) | (df["latitude"] > 90),
    "latitude"
] = np.nan

# Replace invalid longitude values with NaN
df.loc[
    (df["longitude"] < -180) | (df["longitude"] > 180),
    "longitude"
] = np.nan
```

### **Step 28 — Missing-data Markdown cell**

## Missing Data Handling Strategy

Missing values were handled according to the meaning and data type of each field.

- Numeric measurements such as age, number of arrests, and property loss were imputed using the median because the dataset contains extreme values and the median is less sensitive to outliers.
- Latitude and longitude were imputed using the median value within the corresponding city because geographic coordinates are related to location. The overall median was used when a city-level median was unavailable.
- Categorical fields such as gender, race, weapon, case status, and resolution were assigned an explicit `"Unknown"` category where the original value was missing. This avoids introducing a potentially incorrect category through mode imputation.
- `reported_online` was imputed using the mode because it represents a binary field.
- Missing phone numbers were represented as `"Not Provided"` rather than generating artificial contact information.
- Missing notes were represented as `"No notes provided"`.
- Missing incident timestamps were retained as missing because assigning an estimated timestamp would create information that was not present in the original dataset.
- Missing suspect information was retained as `"Unknown"` rather than deleting the incident because an incident can still contain useful information even when suspect details are unavailable.

```python
# Median imputation for numeric columns
numeric_imputation_columns = [
    "suspect_age",
    "victim_age",
    "num_arrests",
    "property_loss_usd"
]

for col in numeric_imputation_columns:
    df[col] = df[col].fillna(df[col].median())
```

```python
# Impute coordinates using city-level medians
for col in ["latitude", "longitude"]:
    city_median = df.groupby("city")[col].transform("median")
    df[col] = df[col].fillna(city_median)

    # Fallback to overall median if still missing
    df[col] = df[col].fillna(df[col].median())
```

```python
unknown_columns = [
    "badge_number",
    "suspect_id",
    "suspect_first_name",
    "suspect_last_name",
    "suspect_gender",
    "suspect_race",
    "victim_id",
    "victim_first_name",
    "victim_last_name",
    "victim_gender",
    "weapon_used",
    "case_status",
    "resolution"
]

for col in unknown_columns:
    df[col] = df[col].fillna("Unknown")
```

```python
df["victim_phone"] = df["victim_phone"].fillna("Not Provided")
df["notes"] = df["notes"].fillna("No notes provided")
```

```python
df["reported_online"] = df["reported_online"].fillna(
    df["reported_online"].mode()[0]
)
```

### **Step 31 — Outlier detection**

```python
outlier_columns = [
    "latitude",
    "longitude",
    "suspect_age",
    "victim_age",
    "num_arrests",
    "property_loss_usd"
]

outlier_results = []

for col in outlier_columns:
    Q1 = df[col].quantile(0.25)
    Q3 = df[col].quantile(0.75)
    IQR = Q3 - Q1

    lower_bound = Q1 - 1.5 * IQR
    upper_bound = Q3 + 1.5 * IQR

    outlier_count = (
        (df[col] < lower_bound) |
        (df[col] > upper_bound)
    ).sum()

    outlier_results.append({
        "Column": col,
        "Q1": Q1,
        "Q3": Q3,
        "IQR": IQR,
        "Lower Bound": lower_bound,
        "Upper Bound": upper_bound,
        "Outlier Count": outlier_count
    })

outlier_report = pd.DataFrame(outlier_results)

outlier_report
```

### **Step 32 — Outlier decision**

## Outlier Treatment

The IQR method was used to identify statistical outliers in numeric variables.

Statistical outliers were not automatically removed because an extreme value does not necessarily represent an incorrect observation. Values were removed or corrected only when they violated logical domain constraints, such as negative ages or geographic coordinates outside their valid ranges.

Valid extreme values, such as unusually high property loss, were retained because they may represent genuine incidents.
